# Predict energy, force, stress, magmom


```python
from google.colab import files
from pymatgen.core import Structure
from chgnet.model.model import CHGNet

# Step 1. 上傳自己的 MASnI3 CIF
uploaded = files.upload()

# Step 2. 讀取上傳的 CIF
filename = next(iter(uploaded))
structure = Structure.from_file(filename)

# Step 3. 確認晶體資訊
print("Crystal formula:", structure.composition.formula)
print("Number of atoms:", len(structure))
print("Elements:", structure.symbol_set)

# Step 4. 載入 CHGNet 預訓練模型
chgnet = CHGNet.load()

prediction = chgnet.predict_structure(structure)

for key, unit in [
    ("energy", "eV/atom"),
    ("forces", "eV/A"),
    ("stress", "GPa"),
    ("magmom", "mu_B"),
]:
    print(f"CHGNet-predicted {key} ({unit}):\n{prediction[key[0]]}\n")
```

# Structure Optimization


```python
from chgnet.model import StructOptimizer

relaxer = StructOptimizer()
```

```python

from chgnet.model import StructOptimizer

# 使用已載入的 CHGNet 模型
relaxer = StructOptimizer(model=chgnet)

# 最佳化 MASnI3 晶體結構
result = relaxer.relax(
    structure,
    verbose=True
)

# 定義最佳化後的晶體結構
relaxed_structure = result["final_structure"]

print("MASnI3 Relaxed Structure:")
print(relaxed_structure)
```

# Molecular Dynamics


```python
from chgnet.model.dynamics import MolecularDynamics

md = MolecularDynamics(
    atoms=structure,
    model=chgnet,
    ensemble="nvt",
    temperature=1000,  # in k
    timestep=2,  # in fs
    trajectory="md_out.traj",
    logfile="md_out.log",
    loginterval=100,
)
md.run(50)  # run a 0.1 ps MD simulation
```

```python
import numpy as np

magmom = np.asarray(prediction["m"])

print("磁矩數量:", len(magmom))
print("磁矩數值:", magmom)
print("最小磁矩:", np.min(magmom))
print("最大磁矩:", np.max(magmom))
print("是否全部為有限數值:", np.isfinite(magmom).all())
```

```python

import plotly.express as px

fig = df_magmom.hist(
    nbins=20,
    sharex=True,
    sharey=True,
    backend="plotly",
    barmode="overlay",
    layout={"title": "MASnI3 Magmom Distribution"}
)

xmin = df_magmom.min().min() - 0.002
xmax = df_magmom.max().max() + 0.002

fig.update_xaxes(
    range=[xmin, xmax],
    title_text="Magnetic moment (μB)"
)

fig.update_yaxes(title_text="Count")
fig.show()
```
